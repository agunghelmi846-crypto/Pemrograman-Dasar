```python
# soal latihan tiga

# Soal 1: Input Iradiasi Matahari dan SoC Baterai

try:
    iradiasi = float(input("Masukkan iradiasi matahari (W/m²): "))
    soc = float(input("Masukkan SoC baterai (%): "))

    if iradiasi >= 0 and 0 <= soc <= 100:
        print("\n=== DATA SISTEM PLTS ===")
        print(f"Iradiasi Matahari : {iradiasi} W/m²")
        print(f"SoC Baterai       : {soc}%")
        print("Input valid!")

    else:
        print("Input tidak valid!")

except ValueError:
    print("Input harus berupa angka!")
```

```python
# Soal 2: Menghitung Estimasi Daya Panel PLTS

try:
    iradiasi = float(input("Masukkan iradiasi matahari (W/m²): "))
    soc = float(input("Masukkan SoC baterai (%): "))

    if iradiasi >= 0 and 0 <= soc <= 100:

        # Menghitung estimasi daya panel
        daya_panel = iradiasi * 0.18 * 10

        print("\n=== HASIL PERHITUNGAN PLTS ===")
        print(f"Iradiasi Matahari : {iradiasi} W/m²")
        print(f"SoC Baterai       : {soc}%")
        print(f"Estimasi Daya Panel: {daya_panel:.2f} Watt")

    else:
        print("Input tidak valid!")

except ValueError:
    print("Input harus berupa angka!")
```

```python
# Soal 3: Menentukan Mode Operasi PLTS

try:
    iradiasi = float(input("Masukkan iradiasi matahari (W/m²): "))
    soc = float(input("Masukkan SoC baterai (%): "))

    if iradiasi >= 0 and 0 <= soc <= 100:

        # Menghitung estimasi daya panel
        daya_panel = iradiasi * 0.18 * 10

        # Menentukan mode operasi PLTS
        if iradiasi > 600 and soc < 80:
            mode = "CHARGING"
            keterangan = "Iradiasi tinggi, baterai belum penuh, prioritaskan pengisian"

        elif iradiasi > 600 and soc >= 80:
            mode = "GRID-EXPORT"
            keterangan = "Baterai sudah penuh, ekspor kelebihan daya ke jaringan"

        elif iradiasi > 200 and soc >= 20:
            mode = "NORMAL"
            keterangan = "Iradiasi dan baterai cukup, operasi mandiri"

        elif iradiasi <= 200 and soc >= 50:
            mode = "BATTERY-ONLY"
            keterangan = "Iradiasi rendah, andalkan baterai"

        else:
            mode = "EMERGENCY"
            keterangan = "Iradiasi rendah dan baterai hampir habis, kurangi beban"

        # Menampilkan hasil
        print("\n=== HASIL OPERASI PLTS ===")
        print(f"Iradiasi Matahari : {iradiasi} W/m²")
        print(f"SoC Baterai       : {soc}%")
        print(f"Daya Panel        : {daya_panel:.2f} Watt")
        print(f"Mode Operasi      : {mode}")
        print(f"Keterangan        : {keterangan}")

    else:
        print("Input tidak valid!")

except ValueError:
    print("Input harus berupa angka!")
```
