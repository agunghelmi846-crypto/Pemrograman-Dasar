### Jawaban Pertanyaan Analisis (Praktikum 2)

1. **Perbedaan `while input != 99` dengan `for i in range(99)`:**
   * **`while input != 99`**: Merupakan perulangan yang **tidak tentu jumlahnya (uncounted loop / conditional loop)**. Perulangan ini akan terus berjalan sampai pengguna memasukkan angka `99`. Kita tidak tahu pasti pada iterasi keberapa program akan berhenti karena sangat bergantung pada input pengguna.
   * **`for i in range(99)`**: Merupakan perulangan yang **sudah ditentukan jumlahnya (counted loop)**. Perulangan ini akan berjalan pasti sebanyak `99` kali (dari indeks `0` hingga `98`), tanpa peduli nilai input apa pun yang dimasukkan di tengah jalan.

2. **Pentingnya `try-except` dalam menerima input dari pengguna:**
   * Pengguna sering kali melakukan kesalahan ketik atau memasukkan jenis data yang tidak sesuai (misalnya memasukkan huruf/karakter saat program meminta angka `int`).
   * Jika tidak menggunakan `try-except`, kesalahan input tersebut akan memicu error (`ValueError`) yang menyebabkan program langsung berhenti atau *crash* secara tiba-tiba.
   * Dengan `try-except`, kita dapat menangani (*handle*) error tersebut dengan baik, menampilkan pesan peringatan kepada pengguna, dan menjaga program agar tetap berjalan.

3. **Definisi *Infinite Loop* beserta contoh kondisinya:**
   * ***Infinite Loop* (perulangan tanpa henti)** adalah situasi di mana kondisi perulangan (biasanya pada `while`) selalu bernilai `True`, sehingga program terus berjalan berulang-ulang tanpa pernah bisa selesai atau keluar dari blok perulangan tersebut secara normal.
   * **Contoh Kondisi Penyebab:**
     1. **Lupa memperbarui variabel kontrol (counter):**
        ```python
        hitung = 1
        while hitung <= 5:
            print(hitung)
            # Lupa menuliskan 'hitung += 1'
        ```
     2. **Kondisi yang selalu bernilai benar secara statis:**
        ```python
        while True:
            print("Ini tidak akan pernah berhenti!")
        ```

### Jawaban Pertanyaan Analisis (Praktikum 3)

1. **Perbedaan `for i in range(len(list))` dengan `for item in list`:**
   * **`for item in list`**:
     * *Fungsi:* Mengakses langsung setiap nilai/elemen di dalam list satu demi satu.
     * *Kapan tepat digunakan:* Ketika kita hanya memerlukan data di dalam list tersebut untuk dibaca (read-only) dan tidak peduli dengan posisi indeksnya.
   * **`for i in range(len(list))`**:
     * *Fungsi:* Mengakses indeks numerik dari setiap elemen di dalam list.
     * *Kapan tepat digunakan:* Ketika kita perlu memodifikasi isi list asli berdasarkan indeksnya, atau ketika posisi/indeks dari elemen tersebut penting untuk logika program kita.

2. **Cara memodifikasi kode Praktikum 3B agar tanda hubung di akhir tidak ikut tercetak:**
   Ada beberapa cara untuk mengatasinya. Salah satu cara yang paling elegan dalam Python adalah menggunakan metode `.join()`:
   ```python
   nama_institusi = "helmi agung nugroho"
   hasil_output = "-".join(nama_institusi)
   print(hasil_output)
   ```
   Atau jika tetap ingin menggunakan perulangan `for`:
   ```python
   nama_institusi = "helmi agung nugroho"
   hasil_output = ""
   for i, karakter in enumerate(nama_institusi):
       if i == len(nama_institusi) - 1:
           hasil_output += karakter  # Karakter terakhir tanpa tanda hubung
       else:
           hasil_output += karakter + "-"
   print(hasil_output)
   ```

3. **Apa yang dikembalikan oleh `enumerate()` jika dicetak langsung?**
   * Jika kita menjalankan `print(enumerate(bulan))`, Python akan mengembalikan sebuah **objek iterator enumerate** beserta alamat memori tempat objek tersebut disimpan, misalnya: `<enumerate object at 0x...>`.
   * Hal ini terjadi karena `enumerate()` adalah generator/iterator yang efisien secara memori. Untuk melihat isinya secara langsung tanpa perulangan, kita harus mengubahnya terlebih dahulu menjadi tipe data terstruktur seperti list, contohnya: `print(list(enumerate(bulan)))`.

````python
# === PRAKTIKUM 4A: break saat menemukan angka -1 ===
data_angka = [10, 20, 30, 50, 60]

print("Memproses data:")
for angka in data_angka:
    if angka == -1:
        print("  → Angka -1 ditemukan! Proses dihentikan.")
        break
    print(f"  Memproses: {angka}")

print("Program selesai.")
````

````python
# === PRAKTIKUM 4B: break saat menemukan angka 0 ===
data_sensor = [45.2, 67.8, 89.1, 55.3, 72.4]

print("Membaca data sensor pembangkit:")
for nilai in data_sensor:
    if nilai == 0:
        print("  ⚠ Pembacaan sensor = 0! Kemungkinan sensor mati. Proses dihentikan.")
        break
    print(f"  Suhu terbaca: {nilai}°C")
````

````python
# === PRAKTIKUM 4C: continue untuk melewati angka negatif ===
data_daya = [120.5, -1, 135.2, 0.0, 98.7, -2, 145.0]

print("Data daya valid (MW):")
for daya in data_daya:
    if daya <= 0:
        print(f"  (Melewati data tidak valid: {daya})")
        continue
    print(f"  Daya: {daya} MW")
````

````python
# === PRAKTIKUM 4D: Penambahan berkelanjutan sampai total > 20 ===
total = 0
bilangan = 1
batas = 20

while total <= batas:
    total += bilangan
    bilangan += 1

print(f"Total pertama yang melebihi {batas}: {total}")
print(f"Bilangan terakhir ditambahkan: {bilangan - 1}")
````

### Jawaban Pertanyaan Analisis (Praktikum 4)

1. **Perbedaan perilaku program pada Langkah 1 (Praktikum 4A) jika `break` diganti dengan `continue`:**
   * **Jika menggunakan `break`**: Ketika program menemukan nilai `-1`, seluruh perulangan akan langsung dihentikan saat itu juga, sehingga data-data setelah `-1` tidak akan pernah diproses.
   * **Jika diganti dengan `continue`**: Ketika program menemukan nilai `-1`, program hanya akan melewati (*skip*) proses cetak untuk angka `-1` tersebut, lalu segera melanjutkan perulangan untuk memeriksa dan memproses data/angka berikutnya hingga selesai.

2. **Mengapa `continue` lebih baik digunakan daripada `if-else` yang panjang dalam kasus Langkah 3 (Praktikum 4C):**
   * **Menghindari *Nested Code* (Indentasi Menumpuk)**: Penggunaan `continue` bertindak sebagai *guard clause*. Kita bisa langsung mengeliminasi data tidak valid di awal blok perulangan. Jika kita menggunakan `if-else` yang panjang, semua logika utama program harus diletakkan di dalam blok `else`, yang membuat indentasi semakin menjorok ke dalam.
   * **Keterbacaan Kode (Readability)**: Kode menjadi jauh lebih bersih dan mudah dipahami karena alur pemrosesan data valid tetap berada di tingkat indentasi utama perulangan tanpa harus terbungkus dalam struktur percabangan yang kompleks.

````python
# === PRAKTIKUM 5C: Tabel Suhu Sensor (Minggu × Hari) ===
nama_unit = ["Unit 1", "Unit 2", "Unit 3"]
suhu_harian = [
    [85.2, 87.1, 86.5, 88.0, 84.9, 85.7, 86.2],   # Unit 1
    [90.1, 91.3, 89.7, 92.0, 90.5, 91.1, 90.8],   # Unit 2
    [78.4, 79.2, 77.8, 80.1, 79.5, 78.9, 79.7],   # Unit 3
]
hari = ["Sen", "Sel", "Rab", "Kam", "Jum", "Sab", "Min"]

print("Laporan Suhu Boiler (°C)")
print(f"{'Unit':<10}", end="")
for h in hari:
    print(f"{h:>6}", end="")
print(f"{'Rata2':>8}")
print("-" * 60)

for i, unit in enumerate(nama_unit):   # outer loop: per unit
    print(f"{unit:<10}", end="")
    total = 0
    for suhu in suhu_harian[i]:         # inner loop: per hari
        print(f"{suhu:>6.1f}", end="")
        total += suhu
    rata2 = total / len(suhu_harian[i])
    print(f"{rata2:>8.1f}")
````

### Jawaban Pertanyaan Analisis (Praktikum 5)

1. **Total iterasi yang terjadi pada Praktikum 5C:**
   * **Outer loop** berjalan sebanyak jumlah elemen dalam `nama_unit`, yaitu **3 kali** (Unit 1, Unit 2, dan Unit 3).
   * **Inner loop** berjalan sebanyak jumlah data suhu per unit, yaitu **7 kali** (7 hari, dari Sen sampai Min).
   * **Total Iterasi Keseluruhan** = $\text{Outer} \times \text{Inner} = 3 \times 7 =$ **21 iterasi**.

2. **Dampak menggunakan nama variabel yang sama pada perulangan luar dan dalam:**
   * Di Python, perulangan `for` **tidak menciptakan local scope baru** (keduanya berada di scope fungsi atau global yang sama).
   * Jika Anda menggunakan nama variabel yang sama (misal `x`) pada loop luar dan loop dalam, akan terjadi **variable shadowing (tumpang tindih)**.
   * Setiap kali loop dalam berjalan, ia akan menimpa (*overwrite*) nilai dari loop luar. Ketika kendali kembali ke loop luar untuk iterasi berikutnya, nilai kontrolnya telah rusak/berubah, yang dapat menyebabkan perulangan tidak berjalan semestinya atau menghasilkan *bug* logika yang sulit dilacak.

3. **Efisiensi Nested Loop dan Alternatifnya:**
   * **Kapan nested loop menjadi tidak efisien?** Nested loop menjadi sangat lambat ketika volume data sangat besar (misalnya ribuan baris dan kolom). Kompleksitas waktunya adalah kuadratik ($O(N \times M)$), yang akan membuat eksekusi program terasa berat dan memakan banyak waktu CPU.
   * **Alternatif dalam Python untuk data tabular:**
     * Menggunakan library **Pandas** (dengan struktur data `DataFrame`). Pandas ditulis menggunakan bahasa C di latar belakang (*vectorized operations*), sehingga manipulasi data tabular dapat dilakukan secara instan tanpa perlu menulis loop bertingkat secara manual.
     * Menggunakan library **NumPy** untuk operasi matematika berbasis matriks/multidimensi secara cepat.
