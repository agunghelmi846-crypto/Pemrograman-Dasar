```python
# langkah 1.1

def cetak_kop(judul):
    """Prosedur: mencetak judul laporan di antara dua garis."""
    garis = "=" * (len(judul) + 4)
    print(garis)
    print("  " + judul)
    print(garis)


cetak_kop("LAPORAN HARIAN UNIT 1")
print("Energi 2 November: 6098 MWh")
print()
cetak_kop("RINGKASAN MINGGUAN")
```

penjelasan
 Baris def cetak_kop(judul) dijalankan satu kali ketika Python membaca dan membuat fungsi. Sementara itu, perintah di dalam fungsi baru dijalankan ketika fungsi tersebut dipanggil. Pada contoh ini, fungsi dipanggil satu kali.

Garis kop kedua memiliki panjang yang menyesuaikan judul yang digunakan. Hal ini karena jumlah garis dihitung berdasarkan banyaknya karakter pada judul, sehingga panjangnya bisa berbeda jika judul diganti.

judul disebut parameter karena menjadi tempat untuk menerima data di dalam fungsi. Sedangkan "RINGKASAN MINGGUAN" disebut argumen karena merupakan nilai yang dimasukkan ketika fungsi dijalankan.

```python
# langkah 1.2

def mwh_ke_gwh(energi_mwh):
    """Ubah MWh menjadi GWh (1 GWh = 1.000 MWh)."""
    return energi_mwh / 1000


def kcal_ke_kj(kalor_kcal):
    """Ubah kcal menjadi kJ (1 kcal = 4,184 kJ)."""
    return kalor_kcal * 4.184


def daya_rata_rata(energi_mwh, jam):
    """Daya rata-rata (MW) = energi (MWh) dibagi lama operasi (jam)."""
    return energi_mwh / jam


print("Energi hari 2  :", mwh_ke_gwh(6098), "GWh")
print("Nilai kalor    :", round(kcal_ke_kj(4400)), "kJ/kg")
print("Daya rata-rata :", round(daya_rata_rata(6098, 24), 1), "MW")
print("Hari 17 (18 jam):", round(daya_rata_rata(jam=18, energi_mwh=3592), 1), "MW")
```

Jika urutannya diubah menjadi daya_rata_rata(24, 6098), hasil perhitungannya juga akan berubah menjadi sekitar 0,00394. Python tidak menampilkan error karena kedua nilai tersebut masih bisa diproses sebagai angka. Perbedaannya hanya terletak pada urutan nilai yang dimasukkan ke fungsi.

```python
# langkah 1.4

def ringkasan(data):
    """Kembalikan (terendah, tertinggi, rata-rata) dari sebuah array."""
    return min(data), max(data), sum(data) / len(data)


energi = [5576, 6098, 6078, 6095, 6074, 6077, 5564]     # MWh, hari 1-7

terendah, tertinggi, rata = ringkasan(energi)
print(f"Terendah  : {terendah} MWh (hari {energi.index(terendah) + 1})")
print(f"Tertinggi : {tertinggi} MWh (hari {energi.index(tertinggi) + 1})")
print("Rata-rata :", round(rata, 1), "MWh")
print("Satu nilai:", ringkasan([7920]))
```

```python
#langkah 2.3

def efisiensi_termal(batubara_t, energi_mwh, nilai_kalor_kcal_kg=4400):
    """Efisiensi termal (%) dari batubara (t) dan energi terkirim (MWh)."""
    heat_rate = batubara_t * nilai_kalor_kcal_kg / energi_mwh       # kcal/kWh
    return 860 / heat_rate * 100


# Hitungan manual: 1 t x 4400 kcal/kg / 2 MWh = 2200 kcal/kWh -> 860 / 2200 = 39,09 %
assert round(efisiensi_termal(1, 2), 2) == 39.09
print("Uji lulus")

e1 = efisiensi_termal(2998, 5576)                          # urutan benar
e2 = efisiensi_termal(5576, 2998)                          # tertukar!
e3 = efisiensi_termal(energi_mwh=5576, batubara_t=2998)    # kata kunci
for e in [e1, e2, e3]:
    print(round(e, 2), "%", "masuk akal" if 25 <= e <= 45 else "TIDAK masuk akal")
```

penjelasan Jika angka 39.09 pada baris assert diubah menjadi 40, program akan menghasilkan AssertionError karena hasil perhitungan fungsi sebenarnya adalah 39.09, bukan 40. Perintah assert digunakan untuk memastikan hasil sesuai dengan nilai yang diharapkan.

e2 tidak masuk akal karena urutan nilai batubara dan energi tertukar, sehingga nilai yang masuk ke dalam perhitungan menjadi tidak sesuai. Python tidak memberikan error karena kedua nilai tersebut tetap berupa angka dan masih bisa diproses oleh program, meskipun urutannya salah.

e3 tetap aman walaupun urutannya dibalik karena menggunakan keyword argument, yaitu dengan menuliskan nama parameter seperti energi_mwh dan batubara_t. Jadi, Python bisa mengenali nilai yang dimasukkan berdasarkan nama parameternya, bukan berdasarkan urutannya.

```python
# langkah 3.1

def heat_rate(batubara_t, energi_mwh, nilai_kalor_kcal_kg=4400):
    """Heat rate (kcal/kWh) dari batubara (t) dan energi terkirim (MWh)."""
    if energi_mwh <= 0:
        raise ValueError("energi_mwh harus positif")
    return batubara_t * nilai_kalor_kcal_kg / energi_mwh


def efisiensi_termal(heat_rate_kcal_kwh):
    """Efisiensi termal (%) dari heat rate. 1 kWh = 860 kcal."""
    return 860 / heat_rate_kcal_kwh * 100


def capacity_factor(energi_mwh, daya_mw=330, jam=24):
    """Capacity factor (%) untuk suatu periode."""
    return energi_mwh / (daya_mw * jam) * 100


def konsumsi_spesifik(batubara_t, energi_mwh):
    """Konsumsi batubara spesifik (t/MWh)."""
    return batubara_t / energi_mwh


# Uji dengan hitungan manual
assert heat_rate(1, 2) == 2200                    # 1 x 4400 / 2
assert round(efisiensi_termal(2150), 2) == 40     # 860 / 2150 = 0,4
assert round(capacity_factor(7920)) == 100        # 330 MW x 24 jam = 7920 MWh
assert konsumsi_spesifik(100, 200) == 0.5         # 100 t / 200 MWh
print("Keempat uji lulus")

hr = heat_rate(3258, 6098)
print(f"Hari 2: HR {hr:,.0f} kcal/kWh, efisiensi {efisiensi_termal(hr):.2f} %, "
      f"CF {capacity_factor(6098):.1f} %, SCC {konsumsi_spesifik(3258, 6098):.3f} t/MWh")
```

penjelasan Jika heat_rate(100, 0) dipanggil, program akan menampilkan ValueError dengan pesan bahwa energi_mwh harus positif. Hal ini terjadi karena fungsi sudah memiliki pemeriksaan untuk mencegah nilai energi nol atau negatif. Cara ini lebih baik daripada langsung membagi dengan nol karena program bisa memberikan pesan kesalahan yang jelas sehingga penyebab masalah lebih mudah diketahui.

Artinya, fungsi bisa dirangkai atau digunakan secara berurutan. Hasil perhitungan dari fungsi heat_rate() dapat langsung dimasukkan sebagai nilai untuk fungsi efisiensi_termal(). Dengan begitu, kita tidak perlu menghitung dan memasukkan hasilnya secara manual.

Setiap fungsi diuji menggunakan angka kecil yang mudah dihitung secara manual agar hasil dari program bisa dibandingkan dengan perhitungan biasa. Dengan cara ini, kesalahan pada rumus atau fungsi lebih mudah ditemukan sebelum digunakan untuk mengolah data asli yang lebih besar.

```python
#langkah 3.2

def status_efisiensi(efisiensi):
    """'BURUK' jika < 36,0 %, 'RENDAH' jika < 36,3 %, selain itu 'NORMAL'."""
    if efisiensi < 36.0:
        return "BURUK"
    if efisiensi < 36.3:
        return "RENDAH"
    return "NORMAL"


for e, harapan in [(35.9, "BURUK"), (36.0, "RENDAH"), (36.29, "RENDAH"), (36.3, "NORMAL"), (36.5, "NORMAL")]:
    assert status_efisiensi(e) == harapan, f"{e} harusnya {harapan}"
    print(f"{e:>5} % -> {status_efisiensi(e)}")
print("Semua batas benar")
```

penjelasanUntuk efisiensi 35,9%, baris if efisiensi < 36.3: tidak dijalankan karena kondisi pertama sudah terpenuhi, yaitu nilai efisiensi kurang dari 36,0%. Fungsi langsung mengembalikan status "BURUK" sehingga pemeriksaan berikutnya dilewati.

Tidak perlu menggunakan elif dan else karena setiap kondisi if sudah diperiksa secara berurutan. Jika kondisi pertama terpenuhi, fungsi langsung mengembalikan hasil. Jika tidak ada kondisi yang terpenuhi, perintah return "NORMAL" menjadi hasil terakhir.

Jika tanda < pertama diubah menjadi <=, pengujian yang gagal adalah nilai 36,0%. Sebab, nilai tersebut akan masuk ke kategori "BURUK", padahal hasil yang diharapkan adalah "RENDAH".

```python
# langkah 3.3

def heat_rate(batubara_t, energi_mwh, nilai_kalor_kcal_kg=4400):
    """Heat rate (kcal/kWh)."""
    return batubara_t * nilai_kalor_kcal_kg / energi_mwh


def efisiensi_termal(heat_rate_kcal_kwh):
    """Efisiensi termal (%) dari heat rate."""
    return 860 / heat_rate_kcal_kwh * 100


def capacity_factor(energi_mwh, daya_mw=330, jam=24):
    """Capacity factor (%) untuk suatu periode."""
    return energi_mwh / (daya_mw * jam) * 100


def status_efisiensi(efisiensi):
    """'BURUK' jika < 36,0 %, 'RENDAH' jika < 36,3 %, selain itu 'NORMAL'."""
    if efisiensi < 36.0:
        return "BURUK"
    if efisiensi < 36.3:
        return "RENDAH"
    return "NORMAL"


def cetak_baris(hari, hr, efi, cf, status):
    """Prosedur: satu baris tabel laporan."""
    print(f"{hari:>4}  {hr:>7,.0f}  {efi:>6.2f}  {cf:>5.1f}  {status}")


# --- Data Unit 1, 14-20 November 2026 ---
hari_awal = 14
energi = [5533, 5566, 6097, 3592, 6059, 6100, 6098]      # MWh
batubara = [2996, 3009, 3257, 1950, 3234, 3261, 3262]    # t

# --- Program utama ---
print(f"{'Hari':>4}  {'HR kcal':>7}  {'Efi %':>6}  {'CF %':>5}  Status")
efisiensi = []
for i in range(len(energi)):
    hr = heat_rate(batubara[i], energi[i])
    efi = efisiensi_termal(hr)
    efisiensi.append(efi)
    cetak_baris(hari_awal + i, hr, efi, capacity_factor(energi[i]), status_efisiensi(efi))

terendah = min(efisiensi)
print(f"\nRata-rata efisiensi : {sum(efisiensi) / len(efisiensi):.2f} %")
print(f"Efisiensi terendah  : {terendah:.2f} % (hari {hari_awal + efisiensi.index(terendah)})")
print(f"CF seminggu         : {capacity_factor(sum(energi), jam=24 * len(energi)):.1f} %")
```

penjelasan Bagian "Program utama" hanya berisi beberapa baris perhitungan dan pemanggilan fungsi, karena semua rumus sudah dibuat sebelumnya di dalam fungsi masing-masing. Jadi, program utama hanya perlu menjalankan fungsi tersebut sesuai data yang tersedia.

cetak_baris disebut prosedur karena tugasnya hanya menampilkan hasil perhitungan dalam bentuk tabel menggunakan print(), bukan mengembalikan nilai untuk digunakan dalam perhitungan lain.

Jika rumus efisiensi harus diubah, cukup satu tempat yang perlu diedit, yaitu pada fungsi efisiensi_termal(). Berbeda dengan skrip salin-tempel di bagian 9 langkah 1, rumus harus diubah di setiap bagian yang menggunakannya. Dengan menggunakan fungsi, perubahan menjadi lebih mudah karena rumus cukup diperbaiki satu kali.

```python
# langkah 3.4
def heat_rate(batubara_t, energi_mwh, nilai_kalor_kcal_kg=4400):
    """Heat rate (kcal/kWh)."""
    return batubara_t * nilai_kalor_kcal_kg / energi_mwh


def efisiensi_termal(heat_rate_kcal_kwh):
    """Efisiensi termal (%) dari heat rate."""
    return 860 / heat_rate_kcal_kwh * 100


def capacity_factor(energi_mwh, daya_mw=330, jam=24):
    """Capacity factor (%) untuk suatu periode."""
    return energi_mwh / (daya_mw * jam) * 100


def laporan_unit(nama, daya_mw, nilai_kalor, energi, batubara):
    """Prosedur: ringkasan KPI satu unit dari array energi (MWh) dan batubara (t)."""
    total_energi = sum(energi)
    hr = heat_rate(sum(batubara), total_energi, nilai_kalor_kcal_kg=nilai_kalor)
    cf = capacity_factor(total_energi, daya_mw=daya_mw, jam=24 * len(energi))
    print(f"{nama}: {len(energi)} hari, HR {hr:,.0f} kcal/kWh, "
          f"efisiensi {efisiensi_termal(hr):.2f} %, CF {cf:.1f} %")


laporan_unit("Unit 1", 330, 4400,
             [5576, 6098, 6078, 6095, 6074, 6077, 5564],
             [2998, 3258, 3250, 3263, 3249, 3246, 2993])

laporan_unit("Unit 3", 100, 4200,
             [2210, 2235, 2190],
             [1270, 1281, 1262])
laporan_unit("Unit 2", 150, 4300,
             [3000, 3100, 3050],
             [1700, 1750, 1720])
```
