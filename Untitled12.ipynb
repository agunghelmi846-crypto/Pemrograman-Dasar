```python
# soal latihan dua

# Soal 1: Input Faktor Daya dan Efisiensi Distribusi

try:
    faktor_daya = float(input("Masukkan faktor daya (0.0 - 1.0): "))
    efisiensi = float(input("Masukkan efisiensi distribusi (%): "))

    if 0.0 <= faktor_daya <= 1.0 and 0 <= efisiensi <= 100:
        print("Input valid!")
        print("Faktor daya:", faktor_daya)
        print("Efisiensi distribusi:", efisiensi, "%")

    else:
        print("Input tidak valid!")

except ValueError:
    print("Input harus berupa angka!")
```

```python
# Soal 2: Menghitung Skor Kualitas Daya

try:
    faktor_daya = float(input("Masukkan faktor daya (0.0 - 1.0): "))
    efisiensi = float(input("Masukkan efisiensi distribusi (%): "))

    if 0.0 <= faktor_daya <= 1.0 and 0 <= efisiensi <= 100:

        skor = faktor_daya * efisiensi

        print("Faktor daya:", faktor_daya)
        print("Efisiensi distribusi:", efisiensi, "%")
        print(f"Skor kualitas daya: {skor:.2f}")

    else:
        print("Input tidak valid!")

except ValueError:
    print("Input harus berupa angka!")
```

```python
# Soal 3: Klasifikasi Kualitas Daya Listrik

try:
    faktor_daya = float(input("Masukkan faktor daya (0.0 - 1.0): "))
    efisiensi = float(input("Masukkan efisiensi distribusi (%): "))

    if 0.0 <= faktor_daya <= 1.0 and 0 <= efisiensi <= 100:

        skor = faktor_daya * efisiensi

        if skor >= 85 and faktor_daya >= 0.90:
            klasifikasi = "PRIMA"
            keterangan = "Sistem dalam kondisi optimal"

        elif skor >= 70 and faktor_daya >= 0.80:
            klasifikasi = "BAIK"
            keterangan = "Sistem memadai, pantau rutin"

        elif skor >= 50 or faktor_daya >= 0.75:
            klasifikasi = "CUKUP"
            keterangan = "Ada ruang perbaikan"

        else:
            klasifikasi = "BURUK"
            keterangan = "Perlu perbaikan segera"

        print("\n=== HASIL KUALITAS DAYA ===")
        print(f"Faktor Daya : {faktor_daya}")
        print(f"Efisiensi   : {efisiensi}%")
        print(f"Skor        : {skor:.2f}")
        print(f"Klasifikasi : {klasifikasi}")
        print(f"Keterangan  : {keterangan}")

    else:
        print("Input tidak valid!")

except ValueError:
    print("Input harus berupa angka!")
```
