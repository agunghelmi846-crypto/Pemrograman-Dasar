```python
angka = int(input("Masukkan sebuah angka: "))

if angka % 5 == 0:
    print("Kelipatan 5")
else:
    print("Bukan Kelipatan 5")
```

```python
tahun = int(input("Masukkan tahun: "))

if tahun % 400 == 0:
    print("Tahun Kabisat")
elif tahun % 4 == 0 and tahun % 100 != 0:
    print("Tahun Kabisat")
else:
    print("Bukan Tahun Kabisat")
```
