```python
# soal ats no 8
# Volume awal tangki
volume = 150

# Perulangan selama volume masih lebih dari 0
while volume > 0:
    volume -= 20
    print(f"Volume saat ini = {volume} liter")

    if volume == 20:
        print("WARNING: Hampir kosong!")

# Pesan setelah tangki kosong
print("Tangki kosong!")
```

penjelasan soal no 8
Variabel volume digunakan untuk menyimpan kapasitas awal tangki sebesar 100 liter. Perulangan while digunakan selama volume masih lebih dari 0. Setiap perulangan, volume dikurangi 20 liter menggunakan operator -=, kemudian ditampilkan. Percabangan if digunakan untuk menampilkan peringatan ketika volume mencapai 20 liter. Setelah volume mencapai 0, perulangan berhenti dan program menampilkan pesan tangki kosong.

```python
data = ["62.5", "58.0", "81.3", "N/A", "66.0"]

jumlah_valid = 0

for i, nilai in enumerate(data, start=1):
    try:
        suhu = float(nilai)
        jumlah_valid += 1

        if suhu > 70.0:
            print(f"Warning: Titik {i} suhu bearing tinggi!")

    except ValueError:
        print(f"Titik {i}: data tidak valid - pemeriksaan dihentikan")
        break

print(f"Jumlah data valid = {jumlah_valid}")
```

penjelasan soal no 10 Perulangan for digunakan untuk memeriksa setiap data suhu, sedangkan enumerate() digunakan untuk memberikan nomor titik mulai dari 1. Fungsi float() mengubah data teks menjadi angka desimal. try-except ValueError digunakan untuk menangani data yang tidak bisa diubah menjadi angka. Jika menemukan data tidak valid, break menghentikan perulangan. Variabel jumlah_valid digunakan untuk menghitung data yang berhasil diperiksa.

```python
# chalange

# Program Pemantauan Konsumsi Listrik Harian
# Program digunakan untuk memeriksa konsumsi listrik selama 5 hari

# List untuk menyimpan data konsumsi listrik (kWh)
konsumsi = [8.5, 12.0, 15.5, 9.0, 18.0]

# Variabel untuk menghitung total konsumsi
total = 0

# Perulangan untuk memeriksa konsumsi setiap hari
for i, energi in enumerate(konsumsi, start=1):
    total += energi

    # Percabangan untuk menentukan status konsumsi
    if energi >= 15:
        status = "TINGGI"
    elif energi >= 10:
        status = "SEDANG"
    else:
        status = "NORMAL"

    print(f"Hari ke-{i}: {energi} kWh - {status}")

# Menghitung rata-rata konsumsi listrik
rata_rata = total / len(konsumsi)

print(f"\nTotal konsumsi = {total} kWh")
print(f"Rata-rata konsumsi = {rata_rata:.2f} kWh")
```

penjelasan bagian challenge Program ini dibuat untuk memantau konsumsi listrik selama lima hari. List digunakan untuk menyimpan data konsumsi listrik dalam satuan kWh. Perulangan for digunakan untuk memeriksa data setiap hari, sedangkan percabangan if-elif-else digunakan untuk menentukan status konsumsi berdasarkan jumlah energi yang digunakan. Variabel total digunakan untuk menghitung keseluruhan konsumsi listrik, kemudian dibagi dengan jumlah data untuk mendapatkan rata-rata konsumsi harian.
