```python
skor = [80, 65, 90, 75, 70]

total = 0                    # akumulator: mulai dari nol
jumlah_lulus = 0             # penghitung = akumulator yang menambah 1
for nilai in skor:
    total = total + nilai    # tambahkan setiap elemen
    if nilai >= 75:
        jumlah_lulus = jumlah_lulus + 1
    print("total sementara:", total)

rata_rata = total / len(skor)
print("Total     :", total)
print("Rata-rata :", rata_rata)
print("Skor >= 75:", jumlah_lulus, "dari", len(skor))
```

```python
energi = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,
]

total = 0
i_maks = 0
i_min = 0
for i in range(len(energi)):
    total = total + energi[i]          # akumulator
    if energi[i] > energi[i_maks]:
        i_maks = i                      # simpan POSISI, bukan hanya nilai
    if energi[i] < energi[i_min]:
        i_min = i

print(f"Energi sebulan : {total:,} MWh")
print(f"Rata-rata      : {total / len(energi):,.0f} MWh/hari")
print(f"Tertinggi      : hari ke-{i_maks + 1} ({energi[i_maks]:,} MWh)")
print(f"Terendah       : hari ke-{i_min + 1} ({energi[i_min]:,} MWh)")
print("Fungsi bawaan  :", sum(energi), max(energi), min(energi))
```

```python
energi = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,
]
hari = list(range(1, 31))
jam_operasi = [24] * 30
jam_operasi[16] = 18
NAMA_HARI = ["Min", "Sen", "Sel", "Rab", "Kam", "Jum", "Sab"]

for i in range(len(energi)):
    if jam_operasi[i] < 24:
        print("Hari", hari[i], ":", jam_operasi[i], "jam,", energi[i], "MWh")

kerja = []
libur = []
for i in range(len(energi)):
    if NAMA_HARI[i % 7] == "Sab" or NAMA_HARI[i % 7] == "Min":
        libur.append(energi[i])
    else:
        kerja.append(energi[i])

print("Rata-rata hari kerja :", round(sum(kerja) / len(kerja)), "MWh")
print("Rata-rata akhir pekan:", round(sum(libur) / len(libur)), "MWh")
```

```python

# Membuat list suhu selama 7 hari
suhu = [28, 30, 29, 32, 31, 27, 33]

print("Data suhu:", suhu)

# 1. Menampilkan 3 hari pertama dan 2 hari terakhir
print("3 hari pertama:", suhu[:3])
print("2 hari terakhir:", suhu[-2:])

# 2. Menambahkan dan mengeluarkan data
suhu.append(34)
print("Setelah append:", suhu)

suhu.pop()
print("Setelah pop:", suhu)

# 3. Mengurutkan suhu tanpa mengubah list asli
print("Suhu terurut:", sorted(suhu))
print("List asli:", suhu)

# 4. Menghitung rata-rata dan suhu tertinggi
rata_rata = sum(suhu) / len(suhu)
tertinggi = max(suhu)
hari_tertinggi = suhu.index(tertinggi) + 1

print("Rata-rata suhu:", rata_rata)
print("Suhu tertinggi:", tertinggi)
print("Suhu tertinggi terjadi pada hari ke-", hari_tertinggi)

# 5. Membuat list suhu di atas rata-rata
di_atas_rata = []

for nilai in suhu:
    if nilai > rata_rata:
        di_atas_rata.append(nilai)

print("Suhu di atas rata-rata:", di_atas_rata)
```

```python
angka = [1, 2, 2, 3]

# Salah: menghapus dari list saat menelusurinya
for x in angka:
    if x == 2:
        angka.remove(x)
print("salah:", angka)          # satu angka 2 masih tersisa!

# Benar: buat list baru hanya dengan yang ingin dipertahankan
angka = [1, 2, 2, 3]
sisa = []
for x in angka:
    if x != 2:
        sisa.append(x)
print("benar:", sisa)
```

```python

# Membuat list skor mahasiswa
skor = [80, 65, 90, 75, 70, 85]

# 1. Mencetak skor pertama dan terakhir
print("Skor pertama:", skor[0])
print("Skor terakhir:", skor[-1])

# 2. Menghitung rata-rata
rata_rata = sum(skor) / len(skor)
print("Rata-rata skor:", rata_rata)

# 3. Menghitung skor di atas rata-rata
jumlah = 0
for nilai in skor:
    if nilai > rata_rata:
        jumlah += 1

print("Jumlah skor di atas rata-rata:", jumlah)

# 4. Mencari posisi skor terendah
terendah = min(skor)
nomor_mahasiswa = skor.index(terendah) + 1
print("Skor terendah:", terendah)
print("Nomor mahasiswa:", nomor_mahasiswa)

# 5. Membuat list baru dengan setiap skor ditambah 5
skor_baru = []

for nilai in skor:
    skor_baru.append(nilai + 5)

print("Skor setelah ditambah 5:", skor_baru)
```

```python

# Data energi selama 30 hari (kWh)
energi = [
    100, 105, 98, 95, 102, 110, 96,
    99, 103, 94, 97, 108, 101, 92,
    104, 100, 96, 109, 98, 95, 107,
    102, 93, 105, 99, 91, 106, 100,
    97, 103
]

# 1. Menghitung total dan rata-rata energi
total = sum(energi)
rata_rata = total / len(energi)

print("Total energi:", total, "kWh")
print("Rata-rata energi:", round(rata_rata, 2), "kWh")

# 2. Mencari tiga hari tertinggi dan terendah
data_hari = []

for i in range(len(energi)):
    data_hari.append((i + 1, energi[i]))

tertinggi = sorted(data_hari, key=lambda x: x[1], reverse=True)
terendah = sorted(data_hari, key=lambda x: x[1])

print("\n3 hari tertinggi:")
for hari, nilai in tertinggi[:3]:
    print("Hari ke-", hari, ":", nilai, "kWh")

print("\n3 hari terendah:")
for hari, nilai in terendah[:3]:
    print("Hari ke-", hari, ":", nilai, "kWh")

# 3. Menghitung jumlah hari di bawah target
target = float(input("\nMasukkan target energi harian: "))

jumlah_bawah = 0
for nilai in energi:
    if nilai < target:
        jumlah_bawah += 1

print("Jumlah hari di bawah target:", jumlah_bawah)

# 4. Rata-rata bergerak 7 hari
batas = 0.97 * rata_rata

print("\nHari dengan rata-rata bergerak di bawah 97% rata-rata bulanan:")

for i in range(len(energi) - 6):
    rata_7 = sum(energi[i:i+7]) / 7

    if rata_7 < batas:
        print("Hari ke-", i + 7, ":", round(rata_7, 2), "kWh")

# 5. Total energi per minggu
print("\nTotal energi per minggu:")

for i in range(0, len(energi), 7):
    minggu = energi[i:i+7]
    nomor_minggu = (i // 7) + 1

    print("Minggu", nomor_minggu, ":", sum(minggu), "kWh")
```
