```python
# soal latihan satu

# Program Menghitung Faktorial

n = int(input("Masukkan bilangan bulat (1-5): "))

faktorial = 1

if 1 <= n <= 5:
    for i in range(1, n + 1):
        faktorial = faktorial * i

    print("Faktorial dari", n, "adalah", faktorial)
else:
    print("Bilangan harus antara 1 sampai 5")
```

```python
# Program Menghitung Faktorial

n = int(input("Masukkan bilangan bulat (1-5): "))

if n == 1:
    faktorial = 1
elif n == 2:
    faktorial = 2 * 1
elif n == 3:
    faktorial = 3 * 2 * 1
elif n == 4:
    faktorial = 4 * 3 * 2 * 1
elif n == 5:
    faktorial = 5 * 4 * 3 * 2 * 1
else:
    faktorial = None

if faktorial is not None:
    print("Faktorial dari", n, "adalah", faktorial)
else:
    print("Bilangan harus antara 1 sampai 5")
```

```python
# Soal 3: Menampilkan Hasil Faktorial

try:
    n = int(input("Masukkan bilangan bulat (1-5): "))

    if 1 <= n <= 5:
        if n == 1:
            faktorial = 1
        elif n == 2:
            faktorial = 2 * 1
        elif n == 3:
            faktorial = 3 * 2 * 1
        elif n == 4:
            faktorial = 4 * 3 * 2 * 1
        else:
            faktorial = 5 * 4 * 3 * 2 * 1

        print("Hasil faktorial dari", n, "adalah", faktorial)

    else:
        print("Input tidak valid! Masukkan angka 1 sampai 5.")

except ValueError:
    print("Input harus berupa angka!")
```
